```python
import hippo
from mucos import MuCOS_score, MuCOS2_score
import pandas as pd
```

### Initialise HIPPO

```python
animal = hippo.HIPPO("2Ab", "2Ab.sqlite")
animal.add_hits(
    "A71EV2A",
    "/Users/nfo24278/Documents/dphil/targets/A71EV2A/metadata.csv",
    "/Users/nfo24278/Documents/dphil/targets/A71EV2A/aligned_files"
    )
```

### Load Reference Pairs

```python
# Load .csv file with pose pairings
# df = pd.read_csv("/Users/nfo24278/Documents/dphil/targets/A71EV2A/derivative_matching/A71EV2A_sucos_vol_max_yes_mb.csv")
df = pd.read_csv("/Users/nfo24278/Documents/dphil/targets/A71EV2A/derivative_matching/A71EV2A_sucos_vol.csv")
df.head()
```

```python
# Get a row from the dataframe as a list
row = df.iloc[1].tolist()

# Remove NaN values from the list
row = [x for x in row if str(x) != 'nan']
row
```

```python

def do_MuCOS(pose):
    reference = row[0]
    inspirations = animal.poses[reference, pose.name]
    inspirations.draw()
    return (
        MuCOS_score(
        inspirations[0].mol,
        inspirations[1].mol,
        draw=False,
        debug=False,
        return_all=True
        ),
        MuCOS2_score(
        inspirations[0].mol,
        inspirations[1].mol,
        draw=False,
        debug=False,
        return_all=True
        )
        )
```

```python
inspirations = animal.poses[row[1:]]
# inspirations.interactive(method="draw2d")
inspirations.interactive(function=do_MuCOS, method=None)
```
