### First pass preparation

Working with the full dataset, we will simply group all fragments by their association to derivatives. This is arbitrary and mostly to get familiar with the data. Filters will need to be subsequently applied.

```python
df = pd.read_csv('/Users/nfo24278/Documents/dphil/targets/A71EV2A/max_A71EV2A_sucos.csv')
grouped = df.groupby('Derivative Name')['Fragment Name'].apply(lambda x: x).reset_index()

# Group by 'Derivative Name' and combine 'Fragment Name' into a single row
result = grouped.groupby('Derivative Name')['Fragment Name'].apply(list).reset_index()

# Exploding the list into separate columns
result_list_exp = pd.DataFrame(result['Fragment Name'].tolist())

# Setting column names for the new DataFrame
result_list_exp.columns = [f'frag_name_{i+1}' for i in range(result_list_exp.shape[1])]

# Joining the new DataFrame with the original ID column
result_df = result[['Derivative Name']].join(result_list_exp)

# Replace any None with NaN if desired (optional, as pandas automatically uses NaN for missing numerical data)
result_df.fillna(value=np.nan, inplace=True)

# Write to csv
# result_df.to_csv('/Users/nfo24278/Documents/dphil/targets/A71EV2A/derivative_matching/A71EV2A_sucos_paired_all.csv', index=False)
# result_df.head(5)
```
